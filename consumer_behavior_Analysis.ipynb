```python
bins=[0,18,25,35,50,100]
labels=['under 18','18-25','26-35','36-50','50+']
```

```python
df['Age_Group']=pd.cut(df['Age'],bins=bins,labels=labels)
```

```python
df['Age_Group'].value_counts()
```

```python
cols_to_fill=['Gender','Occupation_Status','Country_Region','Shop_Most_Where','Preferred_Platform','Preferred_Payment_Method']
```

```python
for col in cols_to_fill:
    df[col]=df[col].fillna("unknown")
```

```python
age_summary=df.groupby("Age_Group",observed=True).agg(avg_satisfaction=("Online_Shopping_Satisfaction_Score","mean"),
avg_trust=("Trust_In_Online_Reviews_Score","mean"),avg_regret=("Regret_After_Purchase_Frequency","mean"),responds=("Response_ID","count")).round(2)
```

```python
age_summary
```

```python
df.columns
```

```python
top_countries=df['Country_Region'].value_counts().head(9).index
```

```python
top_countries
```

```python
country_summary = (
    df[df['Country_Region'].isin(top_countries)]
    .groupby('Country_Region')
    .agg(
        avg_satisfaction=('Online_Shopping_Satisfaction_Score', 'mean'),
         bnpl_regular_rate=(
        'Uses_BNPL_Installments',
        lambda x: (x == "Yes, regularly").mean()
    ),
        responses=('Response_ID', 'count')
    )
    .round(2)
    .sort_values('responses', ascending=False)
)

print(country_summary)
```

```python
country_summary
```

```python
occupation_summary=df.groupby("Occupation_Status").agg(avg_impluse=('Impulse_Purchase_Frequency',"mean"),avg_discount_importance=('Discount_Importance_Score',"mean"),bnpl_reqular_rate=("Uses_BNPL_Installments",lambda x:(x=="Yes, regularly").mean()),respondents=("Response_ID","count")).round(2).sort_values('avg_impluse',ascending=False)
```

```python
occupation_summary
```

```python
pivot_payment=pd.pivot_table(df,index="Shop_Most_Where",columns="Preferred_Payment_Method",values="Response_ID",aggfunc="count",fill_value=0)
```

```python
pivot_payment
```

```python
pivot_payment_pct=pivot_payment.div(pivot_payment.sum(axis=1),axis=0).round(2)
```

```python
pivot_payment_pct
```

```python
pivot_gender_spend=pd.pivot_table(df,index='Gender',columns="Top_Spending_Category",values="Response_ID",aggfunc="count",fill_value=0)
```

```python
pivot_gender_spend
```

```python
pivot_gender_spend_pct=pivot_gender_spend.div(pivot_gender_spend.sum(axis=1),axis=0).round(2)
```

```python
pivot_gender_spend_pct
```

```python
plt.figure(figsize=(14, 6))

sns.histplot(df['Age'].dropna(), bins= 10, kde=True, color="skyblue")

plt.title("Age Distribution of Survey Respondents")
plt.xlabel("Age")
plt.ylabel("Count")

plt.tight_layout()
plt.show()
```

```python
sns.barplot(x=age_summary.index,y=age_summary['avg_satisfaction'],palette="viridis")
plt.title("Average Online Shopping by Age Group",fontweight="bold")
plt.xlabel("Age Group")
plt.ylabel("Average Satisfaction Score(1-18)")
plt.ylim(0,10)
plt.tight_layout()
plt.show()
```

```python
sns.heatmap(pivot_payment_pct,annot=True,cmap="Blues",fmt=".0%",cbar_kws={"label":"Share of Respondents"})
plt.title("Preffered payment method by shoppind Mode",fontweight="bold")
plt.xlabel("Preferred payment method")
plt.ylabel("Shop most where")
plt.tight_layout()
plt.show()
```
